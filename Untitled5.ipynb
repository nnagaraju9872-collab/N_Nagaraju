```python
%%writefile app.py
import streamlit as st

# Page Configuration
st.set_page_config(page_title="My Portfolio", page_icon="💻", layout="wide")

# Header Section
col1, col2 = st.columns([1, 2])
with col1:
    # Optional profile image
    st.image("https://via.placeholder.com/150", width=150)
with col2:
    st.title("Hi, I'm Alex Doe 👋")
    st.subheader("Data Scientist & Python Developer")
    st.write("Passionate about building data-driven web applications and machine learning models.")

st.markdown("---")

# About Me
st.header("About Me")
st.write("""
I specialize in Python development, data analytics, and machine learning algorithms. 
I enjoy solving complex problems and turning data into actionable insights.
""")

# Skills
st.header("Skills")
st.markdown("""
* **Languages:** Python, SQL, R
* **Tools & Libraries:** Pandas, NumPy, Scikit-Learn, Streamlit, Git
* **Domains:** Data Analysis, Predictive Modeling, Web Scraping
""")

st.markdown("---")

# Projects
st.header("Projects")
proj1, proj2 = st.columns(2)

with proj1:
    st.subheader("📊 Stock Price Predictor")
    st.write("A machine learning app that forecasts stock trends using time-series analysis.")
    st.markdown("[View Code on GitHub](#)")

with proj2:
    st.subheader("🔍 E-Commerce Sentiment Analysis")
    st.write("Processed thousands of customer reviews using NLP to evaluate sentiment.")
    st.markdown("[View Code on GitHub](#)")

st.markdown("---")

# Contact Form
st.header("Get In Touch")
with st.form("contact_form"):
    name = st.text_input("Your Name")
    email = st.text_input("Your Email")
    message = st.text_area("Your Message")
    submit = st.form_submit_button("Send")
    
    if submit:
        st.success("Thank you! Your message has been sent.")
```

```python
import subprocess

# Launch the Streamlit app in the background
process = subprocess.Popen(["streamlit", "run", "app.py"])

print("App launched! Open http://localhost:8501 in your browser.")
```
